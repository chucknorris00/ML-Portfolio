# Tensors

Similar to arrays and matrices

Use: encode the inputs and outputs of a model as well as the parameters



```python
import torch
import numpy as np
```

You can create a tensor directly from data


```python
data=[[1,2,3,4],[5,6,7,8]]
x_data=torch.tensor(data)
```

or from a numpy array

```python
array=np.array(data)
x_array=torch.from_numpy(array)
print(f"Data Tensor: \n{x_data}\n")
```

```python
# or from other tensors
x_ones=torch.ones_like(x_data)
print(f"Ones Tensor: \n{x_ones}\n")
x_rand=torch.rand_like(x_data,dtype=torch.float)
print(f"Random Tensor: \n{x_rand}\n")
```

```python
shape=(2,9,)

zeros_tensor=torch.zeros(shape)
print(f"Zeros Tensor: \n{zeros_tensor}\n")
```

### Atributes to a tensor

shape (like above), dtype (data type) or device (where is it running)

```python
tensor=torch.zeros(6,4)
print(tensor.shape)
print(tensor.dtype)
print(tensor.device)
```

### Operations done on a Tensor

For example: Indexing or slicing

```python
tensor=torch.rand(3,4)
print(f'first row: {tensor[0]}')
print(f'first column: {tensor[:,0]}')
print(f'last column: {tensor[:,-1]}')
print(f"first 3 values of last column: {tensor[:3,-1]}")
```

#### Concatenate (Join Tensors)

```python
tensor_1=torch.rand(2,3) 
tensor_2=torch.rand(2,3)
print(tensor_1,'  ')
print(tensor_2)
```

```python

tensor_sum=torch.cat([tensor_1,tensor_2],dim=1)
print(f'Sum Tensor:\n{tensor_sum}')
```
